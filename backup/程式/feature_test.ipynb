```python
import pandas as pd
from user_agents import parse as ua_parse
from sklearn.preprocessing import LabelEncoder

# 讀取前100筆資料
df = pd.read_csv(r'C:\Users\ian32\Downloads\final\data\train.csv', nrows=100)

# 處理 user_agent 欄位
def extract_ua_features(ua_str):
    ua = ua_parse(str(ua_str))
    browser = ua.browser.family or "Unknown"
    os = ua.os.family or "Unknown"
    if ua.is_mobile:
        device_type = "Mobile"
    elif ua.is_tablet:
        device_type = "Tablet"
    elif ua.is_pc:
        device_type = "PC"
    else:
        device_type = "Other"
    return pd.Series([browser, os, device_type])

df[['browser', 'os', 'device_type']] = df['user_agent'].apply(extract_ua_features)

# 標籤編碼
for col in ['browser', 'os', 'device_type']:
    le = LabelEncoder()
    df[col + '_label'] = le.fit_transform(df[col])

# 顯示處理後的結果
print(df[['user_agent', 'browser', 'os', 'device_type', 'browser_label', 'os_label', 'device_type_label']].head(20))
```
